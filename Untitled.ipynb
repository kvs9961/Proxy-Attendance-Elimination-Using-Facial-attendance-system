```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
import sklearn
import os
import datetime
```

```python
import cv2

vid = cv2.VideoCapture(0)
c = 1 

while True:
    suc, img = vid.read()
    if not suc:
        print("Failed to capture frame.")
        break

    # Draw rectangle
    cv2.rectangle(img, (240, 100), (400, 350), (0, 0, 255), 1)

    crp_img = img[100:350, 240:400]

    cv2.imshow("image", img)
    cv2.imshow("cropped_image", crp_img)

    # Wait for key press
    key = cv2.waitKey(1) & 0xFF

    if key == ord("s"):
        cv2.imwrite(r"C:\Users\HP\Documents\datascience projects\facial attendance system\kvs_images\img_{}.jpg".format(c), crp_img)
        print(f"Image saved: img_{c}.jpg")
        c += 1

    if key == ord("q"):\
        break

vid.release()
cv2.destroyAllWindows()
```

```python
class_var = []
feature = []

for folder in os.listdir(r"C:\Users\HP\Documents\datascience projects\facial attendance system\images"):
    for image in os.listdir(r"C:\Users\HP\Documents\datascience projects\facial attendance system\images\{}".format(folder)):
        img = cv2.imread(r"C:\Users\HP\Documents\datascience projects\facial attendance system\images\{}\{}".format(folder, image))
        resized_img = cv2.resize(img, (100, 100)).flatten()
        feature.append(resized_img)
        class_var.append(folder)
```
