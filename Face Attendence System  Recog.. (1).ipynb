```python
vid=cv2.VideoCapture(0)

c=0
while True:
    suc,img=vid.read()
    
    #print(img.shape)
    
    cv2.rectangle(img,(240,100),(480,450),(1,1,1),1)
    
    crop_img=img[100:450,240:480]
    
    cv2.imshow("capture",img)
    cv2.imshow("croped",crop_img)
    
    if cv2.waitKey(1) & 255 ==ord("s"):
        cv2.imwrite(r"C:\Users\LENOVO\Downloads\saxon\saxon{}.jpg".format(c),crop_img)
        print("save was successful")
        c+=1
    
    if cv2.waitKey(1) & 255 == ord("q"):
        break
    
cv2.destroyAllWindows()
```

```python
features=[]
class_label=[]
c=0
for folder in os.listdir(r"C:\Users\HP\Documents\datascience projects\facial attendance system\sample"):
    for images in os.listdir(r"C:\Users\HP\Documents\datascience projects\facial attendance system\sample\{}".format(folder)):
        img=cv2.imread(r"C:\Users\HP\Documents\datascience projects\facial attendance system\sample\{}\{}".format(folder,images))
        features.append(cv2.resize(img,(100,100)).flatten())
        class_label.append(c)
    c+=1
```

```python
data=pd.DataFrame(features)
data=data.astype(np.uint8) # use it if all the columns are not havinf uint8 dtype
```

```python
knn=KNeighborsClassifier(n_neighbors=15)
model=knn.fit(data,class_label)
```

```python
vid=cv2.VideoCapture(0)
while True:
    suc,img=vid.read()
    
    now = datetime.now()

    
    cv2.rectangle(img,(240,100),(480,450),(1,1,1),1)
    
    crop_img=img[100:450,240:480]
    
    xq=cv2.resize(crop_img,(100,100)).flatten() # pre-processing your query point
    xq=xq.reshape((1,30000)) # pre-processing your query point
    
    y_i=model.predict(xq)
    
    cv2.putText(img,now.strftime("%H:%M:%S"),(10,20),cv2.FONT_HERSHEY_COMPLEX,1,(0,0,255),1)
    
    cv2.putText(img,"Face detection",(10,90),cv2.FONT_HERSHEY_COMPLEX,1,(0,255,255),1)
    
    if y_i[0]==0:
        cv2.putText(img,"Nikhil",(100,150),cv2.FONT_HERSHEY_COMPLEX,1,(0,0,255),1)
        
    
        cv2.imshow("capture",img)
    else:
        cv2.putText(img,"Saxon",(100,150),cv2.FONT_HERSHEY_COMPLEX,1,(0,0,255),1)
    
        cv2.imshow("capture",img)
    
    
    
    if cv2.waitKey(1) & 255 == ord("q"):
        break
    
cv2.destroyAllWindows()
```
